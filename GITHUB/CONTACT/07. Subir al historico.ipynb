```python
import datetime
import calendar

# Obtener la fecha actual
hoy = datetime.date.today()

# Calcular el primer día del mes actual y restar un día para obtener el mes anterior
primer_dia_mes_actual = hoy.replace(day=1)
ultimo_dia_mes_anterior = primer_dia_mes_actual - datetime.timedelta(days=1)

# Extraer mes y año del mes anterior
mes_anterior = ultimo_dia_mes_anterior.month + 11
anio_anterior = ultimo_dia_mes_anterior.year - 1

# Asignar variables
MES_SQL_SP_LIQ_ACTUAL = f"{anio_anterior}{mes_anterior:02d}"

# Imprimir resultados
print("Periodo =", MES_SQL_SP_LIQ_ACTUAL)
```

```python
Etiquetas_mes_acutual2 = Etiquetas_mes_acutual2.fillna(0)
Etiquetas_mes_acutual2.replace('', 0, inplace=True)
Etiquetas_mes_acutual2['VALOR_VENTA']=0  # siempre se iguala a cero y luego se rellena
```

CALCULO EL VALOR X VENTA: Lo bueno de este calculo es que solo asigna valor a lo que esta OK, donde 'Etiqueta_NuevMod' esta lleno por algun valor por ende nunca se divide por cero. Y si no alcanzó el piso al dividirse 'Vent_Parc_PRES' el cual es cero, entre 'P_parcial-PRESENCIAL' por ejemplo = 20, pues 0 /20 =0. Por ende tampoco hay error

```python
def asignar_valor_x_venta(row):
    if row['EstadoFinal'] == 'OK':
        if row['TIPO'] == 'Pregrado':
            if row['Etiqueta_NuevMod'] == 'P_parcial-PRESENCIAL':
                return row['Vent_Parc_PRES'] / row['P_parcial-PRESENCIAL']
            elif row['Etiqueta_NuevMod'] == 'P_parcial-VIRTUAL':
                return row['Vent_Parc_VIRT'] / row['P_parcial-VIRTUAL']
            elif row['Etiqueta_NuevMod'] == 'P_Total-PRESENCIAL':
                return row['Vent_Total_PRES'] / row['P_Total-PRESENCIAL']
            elif row['Etiqueta_NuevMod'] == 'P_Total-VIRTUAL':
                return row['Vent_Total_VIRT'] / row['P_Total-VIRTUAL']
        elif row['TIPO'] == 'Postgrado':
            if row['Etiqueta_NuevMod'] == 'P_parcial':
                return row['Vent_Parc_posg'] / row['P_parcial_posg']
            elif row['Etiqueta_NuevMod'] == 'P_Total':
                return row['Vent_Total_posg'] / row['P_Total_posg']
        elif row['TIPO'] == 'Educacion continuada':
            return row['Valor_Vent_Ec'] / row['Ventas_Ec']
    return 0

Etiquetas_mes_acutual2['VALOR_VENTA'] = Etiquetas_mes_acutual2.apply(asignar_valor_x_venta, axis=1)
#--------------------------------------------------------------
#--------------------------------------------------------------
```

```python
Etiquetas_mes_acutual2['VENTA_DEDUCIDA']=0
Etiquetas_mes_acutual2['VENTA_DEDUCIDA']=Etiquetas_mes_acutual2['VALOR_VENTA']-(Etiquetas_mes_acutual2['VALOR_VENTA']*Etiquetas_mes_acutual2['Por_Afectacion'])
```
