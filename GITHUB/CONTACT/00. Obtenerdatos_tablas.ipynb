## GUARDAR EL HISTORICO ANTES DE LIQUIDAR

```python
from sqlalchemy import create_engine, text, inspect
from sqlalchemy.orm import sessionmaker
import pandas as pd
import os
#from datetime import datetime
#CREACION DE LA CONEXION CON BASE DE DATOS SQL SERVER

# Configura los detalles de la conexión
host='172.16.1.33'
port='1433'
database= 'Cun_Repositorio'
schema='Comisiones'

try:
    # Crea la cadena de conexión
   #url = 'mssql+pyodbc://@{host}:{port}/{db}?driver=SQL+Server'.format(host=host, port=port, db=database)
   url = 'mssql+pyodbc://@{host}:{port}/{db}?trusted_connection=yes&driver=SQL+Server'.format(host=host, port=port, db=database)

    # Crear conexion con base de datos
   engine = create_engine(url) 
   print("Conexion a la base de datos realizada") 
    
except Exception as e:
    print("Error:", e)

try:
    # Lista de las tablas específicas
    tables = [
        'Comisiones_Contact.BasePago',
        'Comisiones_Contact.Etiquetas',
        'Comisiones_Contact.EtiquetasReliquidacion',
        'Comisiones.Ventas_Contact',
        'Comisiones_Contact.Afectaciones_Contact',
        'Comisiones_Contact.FIJA_LIQ_ACTUAL_DESGLOS',
        'Comisiones_Contact.PREGR_APROB_CONTAC_MOD2',
        'Comisiones_Contact.POSG_APROB_CONTAC_MOD2',
        'Comisiones_Contact.BasePagoReliquidaciones',
        'Comisiones_Contact.RELIQ_PREGR_APROB_CONTAC_MOD2',
        'Comisiones_Contact.RELIQ_POSG_APROB_CONTAC_MOD2',
        'Comisiones_Contact.RELIQ_SOL_PREG_X',
        'Comisiones_Contact.RELIQ_SOL_POST_X',
        'Comisiones_Contact.RELIQ_DESGLO_BASE'
    ]
    
    # Iterar sobre cada tabla y exportar a CSV
    for table_name in tables:
        query_table = f"SELECT * FROM {table_name}"
        data = pd.read_sql(query_table, engine)
        
        # Guardar los datos de cada tabla en un archivo CSV
        output_directory=r'C:\Users\alejandro_arboledas\OneDrive - Corporación Unificada Nacional de Educación Superior - CUN\Documentos\PYTHON\CONTACT\HISTORICO'
        output_file = os.path.join(output_directory, f"{table_name.replace('Comisiones_Contact.', '').replace('Comisiones.', '')}.csv")
        data.to_csv(output_file, index=False, encoding='UTF-8', sep='|')
        
        # Confirmación de exportación
        print(f"Datos de la tabla '{table_name}' exportados a {output_file}")

except Exception as e:
    print(f"Error: {e}")
finally:
    engine.dispose()
```

```python
from sqlalchemy import create_engine
import pandas as pd
import os
from datetime import datetime

# -----------------------------
# CONFIG: definir PeriodoPago y directorio de salida
# -----------------------------
PeriodoPago = '202610'  # <- ajusta según necesites
output_directory = r'C:\Users\alejandro_arboledas\OneDrive - Corporación Unificada Nacional de Educación Superior - CUN\Documentos\PYTHON\CONTACT\OCTUBRE'

# -----------------------------
# Obtener mes anterior en español
# -----------------------------
meses_es = [
    "enero","febrero","marzo","abril","mayo","junio",
    "julio","agosto","septiembre","octubre","noviembre","diciembre"
]
hoy = datetime.today()
mes_anterior_num = hoy.month - 2 + 1  # shift so january->0 index for previous month
# simpler: if month == 1 previous is 12 (index 11)
if hoy.month == 1:
    mes_anterior_nombre = meses_es[11]
else:
    mes_anterior_nombre = meses_es[hoy.month - 2]

# -----------------------------
# Conexión a la base de datos
# -----------------------------
host='172.16.1.33'
port='1433'
database= 'Cun_Repositorio'

url = f"mssql+pyodbc://@{host}:{port}/{database}?trusted_connection=yes&driver=SQL+Server"

try:
    engine = create_engine(url)
    print("Conexión a la base de datos realizada")
except Exception as e:
    print("Error conexión:", e)
    raise

# -----------------------------
# Consultas (las 6)
# -----------------------------
queries = {
    'BasePago': f"SELECT distinct * FROM Comisiones_Contact.BasePago WHERE Periodo_Pago = '{PeriodoPago}'",
    'Comisiones_Contact_Etiquetas_OK': f"SELECT distinct * FROM Comisiones_Contact.Etiquetas WHERE Periodo_Pago = '{PeriodoPago}' AND EstadoFinal = 'OK'",
    'Comisiones_Contact_Etiquetas_Not_OK': f"SELECT distinct * FROM Comisiones_Contact.Etiquetas WHERE Periodo_Pago = '{PeriodoPago}' AND EstadoFinal <> 'OK'",
    'FIJA_LIQ_ACTUAL_DESGLOS': f"SELECT distinct * FROM Comisiones_Contact.FIJA_LIQ_ACTUAL_DESGLOS WHERE Periodo_Pago = '{PeriodoPago}'",
    'PREGR_APROB_CONTAC_MOD2': f"SELECT distinct * FROM Comisiones_Contact.PREGR_APROB_CONTAC_MOD2 WHERE Periodo_Pago_1 = '{PeriodoPago}'",
    'APROB_CONTAC_MOD2': f"SELECT distinct * FROM Comisiones_Contact.POSG_APROB_CONTAC_MOD2 WHERE Periodo_Pago_1 = '{PeriodoPago}'"
}

# Asegurar carpeta de salida
os.makedirs(output_directory, exist_ok=True)

# -----------------------------
# Ejecutar consultas y generar archivos
# -----------------------------
try:
    # Lecturas para el archivo combinado (las tres primeras)
    df_basepago = pd.read_sql(queries['BasePago'], engine)
    df_et_ok = pd.read_sql(queries['Comisiones_Contact_Etiquetas_OK'], engine)
    df_et_not = pd.read_sql(queries['Comisiones_Contact_Etiquetas_Not_OK'], engine)

    # 1) Archivo combinado con 3 hojas
    archivo_combinado = os.path.join(output_directory, f"Basepago_{mes_anterior_nombre}.xlsx")
    with pd.ExcelWriter(archivo_combinado, engine='openpyxl') as writer:
        # hoja 1: basepago
        df_basepago.to_excel(writer, sheet_name='basepago', index=False)
        # hoja 2: pasaron (etiquetas ok)
        df_et_ok.to_excel(writer, sheet_name='pasaron', index=False)
        # hoja 3: nopasaron (etiquetas no ok)
        df_et_not.to_excel(writer, sheet_name='nopasaron', index=False)
    print(f"Creado: {archivo_combinado}")

    # 2) Las otras 3 consultas: cada una en su propio archivo (manteniendo el nombre original + mes)
    otros = ['FIJA_LIQ_ACTUAL_DESGLOS', 'PREGR_APROB_CONTAC_MOD2', 'APROB_CONTAC_MOD2']
    for key in otros:
        df = pd.read_sql(queries[key], engine)
        nombre_archivo = os.path.join(output_directory, f"{key}_{mes_anterior_nombre}.xlsx")
        df.to_excel(nombre_archivo, index=False)
        print(f"Creado: {nombre_archivo}")

    # (Opcional) Si quieres también guardar las tablas individuales BasePago/OK/NotOK por separado
    # puedes descomentar las siguientes líneas:
    # pd.DataFrame(df_basepago).to_excel(os.path.join(output_directory, f"BasePago_{mes_anterior_nombre}.xlsx"), index=False)
    # pd.DataFrame(df_et_ok).to_excel(os.path.join(output_directory, f"Comisiones_Contact_Etiquetas_OK_{mes_anterior_nombre}.xlsx"), index=False)
    # pd.DataFrame(df_et_not).to_excel(os.path.join(output_directory, f"Comisiones_Contact_Etiquetas_Not_OK_{mes_anterior_nombre}.xlsx"), index=False)

except Exception as e:
    print("Error durante la ejecución/export:", e)
finally:
    engine.dispose()
    print("Conexión cerrada.")
```
