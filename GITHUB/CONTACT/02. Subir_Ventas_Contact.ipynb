# ACA EXTRAIGO FECHA_PAGO_LIQ

```python
def extraccion_recibos_contact(df_cedulas_periodos):    
    para_orden = df_cedulas_periodos['Número_de_Documento'].astype(str).unique()
    valores_unicos_df1_orden = "', '".join(map(str, para_orden))
    valores_unicos_df1_orden = f"'{valores_unicos_df1_orden}'"
    
    para_orden_per = df_cedulas_periodos['Periodo'].astype(str).unique()
    valores_unicos_df1_orden_per = "', '".join(map(str, para_orden_per))
    valores_unicos_df1_orden_per = f"'{valores_unicos_df1_orden_per}'"

    #-----------------------------------------
    host2_1 = '172.16.1.33'
    port2_1 = '1433'
    database2_1 = 'CUN_REPOSITORIO'

    df_RecibCont2_5 = pd.DataFrame()  # evita UnboundLocalError si falla el try

    try:
        url2_1 = (
            f"mssql+pyodbc://@{host2_1}:{port2_1}/{database2_1}"
            "?trusted_connection=yes&driver=SQL+Server"
        )
        engine2_1 = create_engine(url2_1)
        print("✅ Conexión a la base de datos realizada")

        query = f"""
            SELECT DISTINCT 
                doc_alum AS Documento_Estudiante_zoho, 
                PERIODO, 
                ESTADO_PAGO, 
                EST_MATRICULADO, 
                NUEVO, 
                CLASE_ACTUAL, 
                fuerzazoho AS FuerzaComercialFinal, 
                FEC_PAGO_LIQ,
                NOM_PROGRAMA
            FROM (
                SELECT *
                FROM CUN_REPOSITORIO.zoho.base_personas
                WHERE doc_alum IN ({valores_unicos_df1_orden})
                  AND ESTADO_PAGO = 'PAGO'
                  AND EST_MATRICULADO = '1-Activo'
                  AND PERIODO NOT LIKE '%I%'
            ) AS subquery
            WHERE PERIODO IN ({valores_unicos_df1_orden_per});
        """

        df_RecibCont2_5 = pd.read_sql(query, engine2_1)
        print(f"✅ Datos obtenidos: {df_RecibCont2_5.shape[0]} registros")

    except Exception as e:
        print("❌ Error al ejecutar la consulta SQL:", e)

    finally:
        if 'engine2_1' in locals():
            engine2_1.dispose()
            print("🔒 Conexión cerrada")

    #-----------------------------------------
    # Solo continúa si hay datos
    if df_RecibCont2_5.empty:
        print("⚠️ No se obtuvieron datos en la consulta.")
        return df_RecibCont2_5

    # Limpieza y transformación
    df_RecibCont2_5['Documento_Estudiante_zoho'] = (
        df_RecibCont2_5['Documento_Estudiante_zoho'].str.replace(r'\D', '', regex=True)
    )

    df_RecibCont2_5['llave_recb2c'] = (
        df_RecibCont2_5['Documento_Estudiante_zoho'].astype(str)
        + df_RecibCont2_5['PERIODO'].astype(str)
    )

    # Solo con fecha de pago válida
    df_RecibCont2_5 = df_RecibCont2_5[df_RecibCont2_5['FEC_PAGO_LIQ'].notna()]

    # Transformación de fechas
    df_RecibCont2_5['FEC_PAGO_LIQ'] = df_RecibCont2_5['FEC_PAGO_LIQ'].astype(str)
    df_RecibCont2_5['FEC_num'] = df_RecibCont2_5['FEC_PAGO_LIQ'].str.replace('/', '')
    df_RecibCont2_5['FEC_num'] = df_RecibCont2_5['FEC_num'].apply(
        lambda x: x[4:8] + x[2:4] + x[0:2] if len(x) == 8 else x
    )

    # Orden y depuración final
    df_RecibCont2_5 = df_RecibCont2_5.sort_values(
        by=['llave_recb2c', 'FEC_num'], ascending=[True, True]
    ).drop_duplicates(subset='llave_recb2c', keep='last')

    return df_RecibCont2_5
```
