#NORMALIZACION DE DOCUMENTO DE IDENTIDAD

```python
import re
import pandas as pd

def normalizar_identificacion(valor):
    
    # 1️⃣ Validar nulos
    if pd.isna(valor):
        return None
    
    # 2️⃣ Convertir a string
    valor = str(valor)
    
    # 3️⃣ Eliminar todo lo que NO sea número
    valor = re.sub(r'[^0-9]', '', valor)
    
    # 4️⃣ Validar longitud
    if len(valor) == 0:
        return None
    
    if len(valor) > 12:
        return None   # o valor[:10] si quieres truncar
    
    # 5️⃣ Detectar documentos falsos (0000, 1111, etc.)
    if len(set(valor)) == 1:  # todos los dígitos iguales
        return None
    
    # 6️⃣ Evitar números muy cortos
    if len(valor) < 5:
        return None
    
    return valor
```

```python
df_homologaciones_2["Identificacion_Limpia"] = df_homologaciones_2["NumeroDocumento"].apply(normalizar_identificacion)
```

```python
df_homologaciones_3 = df_homologaciones_2[df_homologaciones_2['Identificacion_Limpia'].notna()].copy()
df_homologaciones_3["NumeroDocumento"] = df_homologaciones_3['Identificacion_Limpia']
df_homologaciones_3.drop(columns=['Identificacion_Limpia'], inplace=True)
df_homologaciones_3
```
