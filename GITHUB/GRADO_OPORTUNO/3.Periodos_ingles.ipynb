```python
# Asegurar que las fechas sean tipo datetime respetando el formato "dd/mm/yyyy"
df_final['Fecha Inicio Convocatoria'] = pd.to_datetime(df_final['Fecha Inicio Convocatoria'], dayfirst=True)
df_final['Fecha Fin Convocatoria'] = pd.to_datetime(df_final['Fecha Fin Convocatoria'], dayfirst=True)
df_final['Fecha Inicio Clases'] = pd.to_datetime(df_final['Fecha Inicio Clases'], dayfirst=True)
df_final['Fecha Fin Clases'] = pd.to_datetime(df_final['Fecha Fin Clases'], dayfirst=True)
```

```python
# Obtener la fecha actual
fecha_actual = pd.to_datetime(datetime.now().strftime('%Y-%m-%d'))  # También puede ser datetime.today()

# Función para determinar la marca
def asignar_marca(row):
    if row['Fecha Inicio Convocatoria'] <= fecha_actual <= row['Fecha Fin Convocatoria']:
        return 'EN CONVOCATORIA'
    elif row['Fecha Inicio Clases'] <= fecha_actual <= row['Fecha Fin Clases']:
        return 'PERIODO ACTIVO'
    else:
        return 'PERIODO INACTIVO'

# Aplicar la función al DataFrame
df_final['Estado'] = df_final.apply(asignar_marca, axis=1)
```

```python
from sqlalchemy import create_engine, text, inspect
from sqlalchemy.orm import sessionmaker
import pandas as pd
#from datetime import datetime
#CREACION DE LA CONEXION CON BASE DE DATOS SQL SERVER

# Configura los detalles de la conexión
host='172.16.1.33'
port='1433'
database= 'CUN_REPOSITORIO'

try:
    # Crea la cadena de conexión
   #url = 'mssql+pyodbc://@{host}:{port}/{db}?driver=SQL+Server'.format(host=host, port=port, db=database)
   url = 'mssql+pyodbc://@{host}:{port}/{db}?trusted_connection=yes&driver=SQL+Server'.format(host=host, port=port, db=database)

   
    # Crear conexion con base de datos
   engine = create_engine(url) 
   print("Conexion a la base de datos realizada") 
    
except Exception as e:
    print("Error:", e)
```

```python
periodos_str = ",".join(f"''{p}''" for p in Periodo) # Convertir lista a string con comillas simples

with engine.connect() as connection:
    try:
        # Ajuste de la consulta
        sql_query = text(f'''
            SELECT
                NUM_IDENTIFICACION,
                MAX(COD_PERIODO) as COD_PERIODO,
                COUNT(NUM_IDENTIFICACION) as Niveles
            FROM OPENQUERY([172.16.1.175], 
                'SELECT DISTINCT
                    C.NUM_IDENTIFICACION,
                    A.ID_ALUM_PROGRAMA,
                    A.ID_GRUPO,
                    A.COD_MATERIA,
                    A.COD_PERIODO,
                    D.NOM_MATERIA
                FROM SINU.SRC_ENC_MATRICULA A
                LEFT JOIN SINU.SRC_ALUM_PROGRAMA B ON A.ID_ALUM_PROGRAMA = B.ID_ALUM_PROGRAMA
                LEFT JOIN SINU.SRC_MATERIA D ON A.COD_MATERIA = D.COD_MATERIA
                LEFT JOIN SINU.BAS_TERCERO C ON B.ID_TERCERO = C.ID_TERCERO
                WHERE A.COD_PERIODO IN ({periodos_str})
                ')
            GROUP BY NUM_IDENTIFICACION
        ''')

        # Ejecutar consulta y guardar en DataFrame
        df_activos = pd.read_sql_query(sql_query, engine)

    finally:
        connection.close()
```

#### CRUCE PARA TRAER SI EL PERIODO ESTA ACTIVO O EN CONVOCATORIA

```python
df_cruce = pd.merge(df_activos,df_act_conv[['Periodo','Estado']],how='left',left_on=['COD_PERIODO'],right_on=['Periodo'])
```
