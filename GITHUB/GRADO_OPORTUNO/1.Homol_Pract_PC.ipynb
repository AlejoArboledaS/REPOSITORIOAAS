```python
# Asignar prioridad al resultado (1 para "APROBÓ", 2 para "NO APROBADO", mantener None como está)
ultimo_periodo["Prioridad"] = ultimo_periodo["Resultado"].apply(lambda x: 1 if x == "APROBÓ" else (2 if x == "NO APROBADO" else 3))

# Ordenar el DataFrame por NUM_IDENTIFICACION, ciclo y Prioridad, asegurando que los None en 'ciclo' se mantengan
df = ultimo_periodo.sort_values(
    by=["NUM_IDENTIFICACION", "ciclo", "Prioridad"],
    key=lambda col: col.fillna(float('inf')) if col.name == "Prioridad" else col,
    na_position='last'
)

# Agrupar y conservar los None en la columna 'ciclo'
df_filtrado = (
    df.groupby(["NUM_IDENTIFICACION", "ciclo", "CURSO"], dropna=False, as_index=False)
    .agg(lambda x: x.iloc[0] if not x.isnull().all() else None)
)

# Eliminar la columna de Prioridad si no es necesaria
df_filtrado = df_filtrado.drop(columns=["Prioridad"])
```

```python
# Pivotear para tener una fila por Ciclo y Estudiante, conservando los None en la columna 'ciclo'
result = df_filtrado.pivot_table(
    index=["NUM_IDENTIFICACION", "ciclo"],  # 'ciclo' puede tener None
    columns="CURSO", 
    values="Resultado", 
    aggfunc=lambda x: x.iloc[0] if len(x) > 0 else None  # Conservar None si el grupo está vacío
)

# Rellenar únicamente las celdas vacías en los valores pivotados, sin afectar la columna 'ciclo'
result = result.fillna("NO APROBADO")

# Reiniciar el índice para que sea legible
result = result.reset_index()
```
